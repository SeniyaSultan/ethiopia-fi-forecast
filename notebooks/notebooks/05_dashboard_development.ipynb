```python
# Driver Contribution Chart
import matplotlib.pyplot as plt
import numpy as np
import os

drivers = ['Smartphone', 'Digital ID', 'Gender Gap', 'Agent Network', '4G Coverage']
contributions = [4.2, 2.8, 2.1, 1.5, 1.2]
colors = ['#2E86AB', '#A23B72', '#F18F01', '#C73E1D', '#6A994E']

plt.figure(figsize=(10, 6))
bars = plt.barh(drivers, contributions, color=colors)
plt.xlabel('Contribution to 2027 Forecast (percentage points)')
plt.title('What Drives Financial Inclusion in Ethiopia?')
plt.gca().invert_yaxis()

# Add value labels
for i, (bar, val) in enumerate(zip(bars, contributions)):
    plt.text(val + 0.1, bar.get_y() + bar.get_height()/2, f'+{val}pp', 
             va='center', fontweight='bold')

plt.tight_layout()
# ensure output directory exists before saving
save_path = 'reports/figures/driver_contribution.png'
os.makedirs(os.path.dirname(save_path), exist_ok=True)
plt.savefig(save_path, dpi=300, bbox_inches='tight')
plt.show()
```
